## 2.6 Kernels and execution order

The kernel remembers every variable you create, in the order you **run**
cells, not the order they appear on the page.

```python
price = 4.50
```

```python
cups = 20
cups * price
```

Now change the price. In a live notebook you would edit the first cell and
rerun it, then rerun the calculation. Here the calculation is repeated below
so you can see the new result.

```python
price = 4.75
```

```python
cups = 20
cups * price
```

### Hidden state

`del` removes a variable from the kernel's memory, just as restarting the
kernel or deleting the cell that created it (and then restarting) would. The
next cell fails with a `NameError`. It is tagged `raises-exception`, so
**Run All** carries on past it.

```python
del price
cups * price
```

## 2.7 Magic commands

Lines starting with `%` (line magics) or `%%` (cell magics) are instructions
to Jupyter itself, not Python.

`%pwd` prints the working directory, the folder the notebook runs in.

```python
%pwd
```

`%who` lists the variables you have created; `%whos` adds details.

```python
%who
```

```python
%whos
```

`%timeit` runs a line many times and reports how long it takes.

```python
%timeit sales["revenue"].sum()
```

`%%time` must be the first line of a cell. It times the whole cell once. Here
we add up revenue with a plain Python loop (Chapter 3 explains loops).

```python
%%time
loop_total = 0
for value in sales["revenue"]:
    loop_total += value
print(round(loop_total, 2))
```

```python
%timeit sum(sales["revenue"])
```

`!` runs a shell command. `head -n 4` shows the first four lines of the raw
CSV file. It works on macOS and Linux; on Windows use
`!type ..\data\bean_there_sales.csv` instead (it prints the whole file).

```python
!head -n 4 ../data/bean_there_sales.csv
```

`%lsmagic` lists every magic command available.

```python
%lsmagic
```

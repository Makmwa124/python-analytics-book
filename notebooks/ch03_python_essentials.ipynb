## 3.9 Functions

`def` packages steps under a name so you can reuse them.

````python
def line_revenue(quantity, unit_price):
    """Return the revenue for one product line, rounded to cents."""
    return round(quantity * unit_price, 2)


line_revenue(53, 4.50)
````

````python
line_revenue(23, prices["Blueberry Muffin"])
````

````python
def line_revenue(quantity, unit_price, discount=0.0):
    """Return revenue after an optional discount (0.10 means 10%)."""
    return round(quantity * unit_price * (1 - discount), 2)


print(line_revenue(53, 4.50))
print(line_revenue(53, 4.50, discount=0.10))
````

````python
def day_label(quantity, busy=60, quiet=40):
    """Label a day's sales as Busy, Normal or Quiet."""
    if quantity >= busy:
        return "Busy"
    elif quantity >= quiet:
        return "Normal"
    else:
        return "Quiet"


for day, qty in zip(days, downtown_lattes):
    print(day, qty, day_label(qty))
````

````python
def summarize(values):
    """Return the total, mean and maximum of a list of numbers."""
    total = sum(values)
    return total, total / len(values), max(values)


week_total, week_mean, week_max = summarize(downtown_lattes)
print(week_total, round(week_mean, 1), week_max)
````

````python
def print_total(values):
    print(sum(values))


result = print_total(downtown_lattes)
print(result)
````

## 3.13 Putting it together: the sales file in plain Python

````python
import csv

revenue_by_store = {}
row_count = 0

with open("../data/bean_there_sales.csv", newline="") as f:
    for row in csv.DictReader(f):
        store = row["store"]
        amount = float(row["revenue"])
        revenue_by_store[store] = revenue_by_store.get(store, 0) + amount
        row_count += 1

print(row_count, "rows read")
for store, amount in revenue_by_store.items():
    print(f"{store:<12} ${amount:>10,.2f}")
````

````python
best = max(revenue_by_store, key=revenue_by_store.get)
best
````
